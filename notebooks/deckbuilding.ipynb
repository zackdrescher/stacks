```python
from stacks.parsing.arena import parse_arena_deck_file
from stacks.parsing.csv import parse_csv_collection_file
```

```python
amulet_titan = parse_arena_deck_file('data/decks/amulet_titan.arena')
belcher = parse_arena_deck_file('data/decks/belcher.arena')
b_midrange = parse_arena_deck_file('data/decks/b_midrange.arena')
jeksai = parse_arena_deck_file('data/decks/jeksai.arena')

graaz = parse_arena_deck_file('data/decks/graaz.arena')
krenko = parse_arena_deck_file('data/decks/krenko.arena')
tg_adds = parse_arena_deck_file('data/decks/t+g_adds.arena')

amulet_titan.add_tag("amulet_titan")
belcher.add_tag("belcher")
b_midrange.add_tag("b_midrange")
jeksai.add_tag("jeksai")

graaz.add_tag("graaz")
krenko.add_tag("krenko")
tg_adds.add_tag("tg_adds")
```

```python
commander_cards = graaz.union(krenko).union(tg_adds)
commander_cards.add_tag("commander")
```

```python
modern_cards = amulet_titan.union(belcher).union(b_midrange).union(jeksai)
modern_cards.add_tag("modern")
```

```python
whishlist = commander_cards.union(modern_cards)

whishlist.add_tag("wishlist")
```

```python
collection = parse_csv_collection_file('data/export_simple_1751163593.csv')

collection.add_tag("collection")
```

```python
cards_i_have, cards_i_dont_have = whishlist.split_on_overlap(collection)
```
